```python
Dataset_ROOT = os.path.join(
    DATASET_PATH,
    "Dataset"
)

train_dir = os.path.join(
    Dataset_ROOT,
    "Train"
)

validation_dir = os.path.join(
    Dataset_ROOT,
    "Validation"
)

test_dir = os.path.join(
    Dataset_ROOT,
    "Test"
)

print("Train:", train_dir)
print("Validation:", validation_dir)
print("Test:", test_dir)
```

```python
import tensorflow as tf

IMG_SIZE_256 = (256, 256)
BATCH_SIZE_256 = 32

train_ds_256 = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    labels="inferred",
    label_mode="binary",
    image_size=IMG_SIZE_256,
    batch_size=BATCH_SIZE_256,
    shuffle=True,
    seed=42
)

validation_ds_256 = tf.keras.utils.image_dataset_from_directory(
    validation_dir,
    labels="inferred",
    label_mode="binary",
    image_size=IMG_SIZE_256,
    batch_size=BATCH_SIZE_256,
    shuffle=False
)

test_ds_256 = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    labels="inferred",
    label_mode="binary",
    image_size=IMG_SIZE_256,
    batch_size=BATCH_SIZE_256,
    shuffle=False
)

# Improve input pipeline
AUTOTUNE = tf.data.AUTOTUNE

train_ds_256 = train_ds_256.prefetch(AUTOTUNE)
validation_ds_256 = validation_ds_256.prefetch(AUTOTUNE)
test_ds_256 = test_ds_256.prefetch(AUTOTUNE)

print("\nâœ… 256Ã—256 datasets ready")
print("Image size:", IMG_SIZE_256)
print("Batch size:", BATCH_SIZE_256)
print("Classes: ['Fake', 'Real']")
```

```python
from tensorflow.keras import layers, models

model_256 = models.Sequential([

    layers.Input(shape=(256, 256, 3)),

    layers.Rescaling(1./255),

    # Block 1
    layers.Conv2D(
        32,
        (3, 3),
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    # Block 3
    layers.Conv2D(
        128,
        (3, 3),
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    # Block 4
    layers.Conv2D(
        256,
        (3, 3),
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    # Feature aggregation
    layers.GlobalAveragePooling2D(),

    # Classification
    layers.Dense(
        256,
        activation="relu"
    ),

    layers.Dropout(0.5),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])

model_256.summary()
```

```python
model_256.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )
    ]
)

print("âœ… 256Ã—256 model compiled")
```

```python
# ============================================
# 256Ã—256 CHECKPOINTS
# ============================================

epoch_checkpoint_256 = tf.keras.callbacks.ModelCheckpoint(
    filepath=os.path.join(
        CHECKPOINT_DIR_256,
        "cnn_256_epoch_{epoch:02d}.keras"
    ),
    save_best_only=False,
    save_weights_only=False,
    verbose=1
)

best_checkpoint_256 = tf.keras.callbacks.ModelCheckpoint(
    filepath=os.path.join(
        CHECKPOINT_DIR_256,
        "cnn_256_best.keras"
    ),
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)

early_stopping_256 = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

reduce_lr_256 = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.2,
    patience=2,
    min_lr=1e-7,
    verbose=1
)

print("âœ… Checkpoint and training callbacks ready")
```

```python
import os
import json
import numpy as np
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

BEST_MODEL_256 = "/content/drive/MyDrive/DeepFake_Detection/experiments/cnn_256/checkpoints/cnn_256_best.keras"

model_256_best = tf.keras.models.load_model(BEST_MODEL_256)

print("Best 256Ã—256 model loaded successfully.")
```

```python
test_results_256 = model_256_best.evaluate(
    test_ds_256,
    verbose=1
)

print("\nKeras test results:")
for name, value in zip(model_256_best.metrics_names, test_results_256):
    print(f"{name}: {value:.4f}")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
import numpy as np

# Get predictions
y_prob_256 = model_256_best.predict(test_ds_256, verbose=1).ravel()

# Convert probabilities to classes
y_pred_256 = (y_prob_256 >= 0.5).astype(int)

# True labels
y_true_256 = np.concatenate([
    y.numpy() for x, y in test_ds_256
])

# Metrics
accuracy_256 = accuracy_score(y_true_256, y_pred_256)
precision_256 = precision_score(y_true_256, y_pred_256)
recall_256 = recall_score(y_true_256, y_pred_256)
f1_256 = f1_score(y_true_256, y_pred_256)
auc_256 = roc_auc_score(y_true_256, y_prob_256)

print("========== 256Ã—256 TEST RESULTS ==========")
print(f"Accuracy : {accuracy_256:.4f} ({accuracy_256*100:.2f}%)")
print(f"Precision: {precision_256:.4f} ({precision_256*100:.2f}%)")
print(f"Recall   : {recall_256:.4f} ({recall_256*100:.2f}%)")
print(f"F1 Score : {f1_256:.4f}")
print(f"ROC-AUC  : {auc_256:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_true_256,
        y_pred_256,
        target_names=["Fake", "Real"]
    )
)

print("\nConfusion Matrix:")
cm_256 = confusion_matrix(y_true_256, y_pred_256)
print(cm_256)
```
