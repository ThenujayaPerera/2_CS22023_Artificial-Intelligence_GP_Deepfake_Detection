```python
DATASET_PATH = os.path.join(path, "Dataset")

train_dir = os.path.join(DATASET_PATH, "Train")
validation_dir = os.path.join(DATASET_PATH, "Validation")
test_dir = os.path.join(DATASET_PATH, "Test")

print("Dataset exists:", os.path.exists(DATASET_PATH))
```

```python
IMG_SIZE_128 = (128, 128)
BATCH_SIZE = 32

train_ds_128 = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    labels="inferred",
    label_mode="binary",
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=42
)

validation_ds_128 = tf.keras.utils.image_dataset_from_directory(
    validation_dir,
    labels="inferred",
    label_mode="binary",
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_ds_128 = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    labels="inferred",
    label_mode="binary",
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    shuffle=False
)

AUTOTUNE = tf.data.AUTOTUNE

train_ds_128 = train_ds_128.prefetch(AUTOTUNE)
validation_ds_128 = validation_ds_128.prefetch(AUTOTUNE)
test_ds_128 = test_ds_128.prefetch(AUTOTUNE)

print("Classes:", train_ds_128.class_names)
print("Image size:", IMG_SIZE_128)
print("Batch size:", BATCH_SIZE)
print("âœ… 128Ã—128 datasets ready.")
```

```python
print("Classes: ['Fake', 'Real']")
print("Image size:", IMG_SIZE_128)
print("Batch size:", BATCH_SIZE)
print("âœ… 128Ã—128 datasets ready.")
```

```python
model_128 = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    layers.Rescaling(1./255),

    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(256, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.GlobalAveragePooling2D(),

    layers.Dense(256, activation="relu"),
    layers.Dropout(0.5),

    layers.Dense(1, activation="sigmoid")
])

model_128.summary()
```

```python
model_128.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

print("âœ… Model compiled.")
```

```python
config_128 = {
    "experiment": "Experiment 3 - CNN 128x128",
    "dataset": "manjilkarki/deepfake-and-real-images",
    "input_size": "128x128x3",
    "train_images": 140002,
    "validation_images": 39428,
    "test_images": 10905,
    "classes": {
        "Fake": 0,
        "Real": 1
    },
    "batch_size": 32,
    "epochs": 10,
    "augmentation": False,
    "optimizer": "Adam",
    "learning_rate": 0.001,
    "loss": "binary_crossentropy",
    "architecture": [
        "Conv2D(32)",
        "MaxPooling2D",
        "Conv2D(64)",
        "MaxPooling2D",
        "Conv2D(128)",
        "MaxPooling2D",
        "Conv2D(256)",
        "MaxPooling2D",
        "GlobalAveragePooling2D",
        "Dense(256)",
        "Dropout(0.5)",
        "Dense(1, sigmoid)"
    ]
}

config_path = os.path.join(
    EXP128_DIR,
    "cnn_128_model_config.json"
)

with open(config_path, "w") as f:
    json.dump(config_128, f, indent=4)

print("Configuration saved:")
print(config_path)
```

```python
epoch_checkpoint = tf.keras.callbacks.ModelCheckpoint(
    filepath=os.path.join(
        CHECKPOINT_DIR,
        "cnn_128_epoch_{epoch:02d}.keras"
    ),
    save_best_only=False,
    save_weights_only=False,
    verbose=1
)
```

```python
best_checkpoint = tf.keras.callbacks.ModelCheckpoint(
    filepath=os.path.join(
        CHECKPOINT_DIR,
        "cnn_128_best.keras"
    ),
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)
```

```python
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.2,
    patience=2,
    min_lr=1e-7,
    verbose=1
)
```

```python
history_128 = model_128.fit(
    train_ds_128,
    validation_data=validation_ds_128,
    epochs=10,
    callbacks=[
        epoch_checkpoint,
        best_checkpoint,
        early_stopping,
        reduce_lr
    ]
)
```

```python
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# ============================================
# GET TRUE LABELS AND PREDICTIONS
# ============================================

y_true = []
y_prob = []

for images, labels in test_ds_128:
    predictions = best_model_128.predict(images, verbose=0)

    y_true.extend(labels.numpy().flatten())
    y_prob.extend(predictions.flatten())

y_true = np.array(y_true).astype(int)
y_prob = np.array(y_prob)

# Convert probability to class
y_pred = (y_prob >= 0.5).astype(int)

# ============================================
# CALCULATE METRICS
# ============================================

accuracy = accuracy_score(y_true, y_pred)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_true,
    y_prob
)

print("========================================")
print("128Ã—128 CNN â€” TEST RESULTS")
print("========================================")

print(f"Accuracy : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision: {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall   : {recall:.4f} ({recall*100:.2f}%)")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=["Fake", "Real"],
        digits=4
    )
)

print("\n========================================")
print("CONFUSION MATRIX")
print("========================================")

cm = confusion_matrix(y_true, y_pred)

print(cm)
```
