```python
model_128_deeper_bn = models.Sequential([

    layers.Input(
        shape=(128, 128, 3)
    ),

    # Normalize pixels
    layers.Rescaling(1./255),


    # =====================================================
    # BLOCK 1
    # =====================================================

    layers.Conv2D(
        32,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.Conv2D(
        32,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # =====================================================
    # BLOCK 2
    # =====================================================

    layers.Conv2D(
        64,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.Conv2D(
        64,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # =====================================================
    # BLOCK 3
    # =====================================================

    layers.Conv2D(
        128,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.Conv2D(
        128,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # =====================================================
    # BLOCK 4
    # =====================================================

    layers.Conv2D(
        256,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.Conv2D(
        256,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # =====================================================
    # BLOCK 5
    # =====================================================

    layers.Conv2D(
        256,
        (3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # =====================================================
    # CLASSIFIER
    # =====================================================

    layers.GlobalAveragePooling2D(),

    layers.Dense(
        256
    ),

    layers.BatchNormalization(),

    layers.Activation("relu"),

    layers.Dropout(
        0.4
    ),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])
```

```python
import os
import numpy as np
import tensorflow as tf

EXP8_MODEL_PATH = (
    "/content/drive/MyDrive/DeepFake_Detection/"
    "experiments/cnn_128_deeper_batchnorm/"
    "checkpoints/cnn_128_deeper_bn_best.keras"
)

model_8 = tf.keras.models.load_model(EXP8_MODEL_PATH)

print("Experiment 8 loaded successfully.")
```

```python
import kagglehub

dataset_path = kagglehub.dataset_download(
    "manjilkarki/deepfake-and-real-images"
)

DATASET_DIR = os.path.join(dataset_path, "Dataset")

VAL_DIR = os.path.join(DATASET_DIR, "Validation")

IMG_SIZE = (128, 128)
BATCH_SIZE = 32

val_ds_8 = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("Class names:", val_ds_8.class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

val_ds_8 = val_ds_8.prefetch(AUTOTUNE)
```

```python
# True validation labels
y_val_true_8 = np.concatenate([
    y.numpy().ravel()
    for _, y in val_ds_8
])

# Predicted probabilities
y_val_prob_8 = model_8.predict(
    val_ds_8,
    verbose=1
).ravel()

print("Validation images:", len(y_val_true_8))
print("Predictions:", len(y_val_prob_8))
```

```python
print("Minimum probability:", y_val_prob_8.min())
print("Maximum probability:", y_val_prob_8.max())
print("Mean probability:", y_val_prob_8.mean())
```

```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

thresholds = np.arange(0.20, 0.81, 0.05)

threshold_results_8 = []

for threshold in thresholds:

    # Predict
    y_val_pred = (
        y_val_prob_8 >= threshold
    ).astype(int)

    # Confusion matrix
    cm = confusion_matrix(
        y_val_true_8,
        y_val_pred
    )

    tn, fp, fn, tp = cm.ravel()

    # Overall accuracy
    accuracy = accuracy_score(
        y_val_true_8,
        y_val_pred
    )

    # Real class metrics
    real_precision = precision_score(
        y_val_true_8,
        y_val_pred,
        pos_label=1,
        zero_division=0
    )

    real_recall = recall_score(
        y_val_true_8,
        y_val_pred,
        pos_label=1,
        zero_division=0
    )

    real_f1 = f1_score(
        y_val_true_8,
        y_val_pred,
        pos_label=1,
        zero_division=0
    )

    # Fake class metrics
    fake_recall = tn / (tn + fp)

    fake_precision = tn / (tn + fn)

    fake_f1 = (
        2 * fake_precision * fake_recall /
        (fake_precision + fake_recall)
        if (fake_precision + fake_recall) > 0
        else 0
    )

    # Macro F1
    macro_f1 = (real_f1 + fake_f1) / 2

    threshold_results_8.append([
        threshold,
        accuracy,
        real_precision,
        real_recall,
        real_f1,
        fake_precision,
        fake_recall,
        fake_f1,
        macro_f1,
        tn,
        fp,
        fn,
        tp
    ])
```

```python
import pandas as pd

threshold_df_8 = pd.DataFrame(
    threshold_results_8,
    columns=[
        "Threshold",
        "Accuracy",
        "Real Precision",
        "Real Recall",
        "Real F1",
        "Fake Precision",
        "Fake Recall",
        "Fake F1",
        "Macro F1",
        "TN",
        "FP",
        "FN",
        "TP"
    ]
)

display(
    threshold_df_8.round(4)
)
```

```python
# ==========================================
# CELL 1 â€” LOAD EXPERIMENT 8 MODEL
# ==========================================

from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("TensorFlow:", tf.__version__)

# Experiment 8
MODEL_PATH = (
    "/content/drive/MyDrive/DeepFake_Detection/"
    "experiments/cnn_128_deeper_batchnorm/"
    "checkpoints/cnn_128_deeper_bn_best.keras"
)

model_8 = tf.keras.models.load_model(MODEL_PATH)

print("\nExperiment 8 model loaded successfully.")
print("Model:", MODEL_PATH)
```

```python
# ==========================================
# CELL 2 â€” LOAD VALIDATION DATA
# ==========================================

import kagglehub

# Download / locate dataset
dataset_path = kagglehub.dataset_download(
    "manjilkarki/deepfake-and-real-images"
)

print("Dataset path:")
print(dataset_path)

VAL_DIR = os.path.join(
    dataset_path,
    "Dataset",
    "Validation"
)

IMG_SIZE = (128, 128)
BATCH_SIZE = 32

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("\nClass names:", val_ds.class_names)
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
```

```python
# ==========================================
# CELL 3 â€” GET VALIDATION PREDICTIONS
# ==========================================

y_val = []
y_val_prob = []

for images, labels in val_ds:

    # Model probability of Real
    probs = model_8.predict(
        images,
        verbose=0
    ).ravel()

    y_val.extend(labels.numpy())
    y_val_prob.extend(probs)

y_val = np.array(y_val)
y_val_prob = np.array(y_val_prob)

print("Validation images :", len(y_val))
print("Predictions       :", len(y_val_prob))

print("\nProbability statistics")
print("Minimum probability:", y_val_prob.min())
print("Maximum probability:", y_val_prob.max())
print("Mean probability   :", y_val_prob.mean())
```

```python
# ==========================================
# CELL 4 â€” BALANCED THRESHOLD ANALYSIS
# ==========================================

thresholds = np.arange(0.20, 0.81, 0.05)

results = []

for threshold in thresholds:

    # Probability >= threshold â†’ Real (1)
    # Probability < threshold  â†’ Fake (0)
    y_pred = (y_val_prob >= threshold).astype(int)

    # Confusion matrix
    tn, fp, fn, tp = confusion_matrix(
        y_val,
        y_pred,
        labels=[0, 1]
    ).ravel()

    # Real = class 1
    real_precision = precision_score(
        y_val,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    real_recall = recall_score(
        y_val,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    real_f1 = f1_score(
        y_val,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    # Fake = class 0
    fake_precision = precision_score(
        y_val,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    fake_recall = recall_score(
        y_val,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    fake_f1 = f1_score(
        y_val,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    macro_f1 = f1_score(
        y_val,
        y_pred,
        average="macro",
        zero_division=0
    )

    accuracy = accuracy_score(
        y_val,
        y_pred
    )

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy,

        "Real Precision": real_precision,
        "Real Recall": real_recall,
        "Real F1": real_f1,

        "Fake Precision": fake_precision,
        "Fake Recall": fake_recall,
        "Fake F1": fake_f1,

        "Macro F1": macro_f1,

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })


results_df = pd.DataFrame(results)

display(
    results_df.round(4)
)
```

```python
# ==========================================
# CELL 2: TEST PREDICTIONS
# ==========================================

y_test = []
y_test_prob = []

for images, labels in test_ds:

    probs = model_8.predict(
        images,
        verbose=0
    ).ravel()

    y_test.extend(labels.numpy())
    y_test_prob.extend(probs)

y_test = np.array(y_test)
y_test_prob = np.array(y_test_prob)

print("Test images:", len(y_test))
print("Predictions:", len(y_test_prob))

print("\nProbability statistics:")
print("Minimum:", y_test_prob.min())
print("Maximum:", y_test_prob.max())
print("Mean   :", y_test_prob.mean())
```

```python
# ==========================================
# CELL 4: FINAL TEST METRICS
# ==========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Confusion matrix
tn, fp, fn, tp = confusion_matrix(
    y_test,
    y_test_pred,
    labels=[0, 1]
).ravel()

# Overall
accuracy = accuracy_score(
    y_test,
    y_test_pred
)

# Real = class 1
real_precision = precision_score(
    y_test,
    y_test_pred,
    pos_label=1,
    zero_division=0
)

real_recall = recall_score(
    y_test,
    y_test_pred,
    pos_label=1,
    zero_division=0
)

real_f1 = f1_score(
    y_test,
    y_test_pred,
    pos_label=1,
    zero_division=0
)

# Fake = class 0
fake_precision = precision_score(
    y_test,
    y_test_pred,
    pos_label=0,
    zero_division=0
)

fake_recall = recall_score(
    y_test,
    y_test_pred,
    pos_label=0,
    zero_division=0
)

fake_f1 = f1_score(
    y_test,
    y_test_pred,
    pos_label=0,
    zero_division=0
)

# Balanced metric
macro_f1 = f1_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

# ROC-AUC uses probabilities, NOT thresholded predictions
roc_auc = roc_auc_score(
    y_test,
    y_test_prob
)

print("=" * 65)
print("EXPERIMENT 8 â€” FINAL TEST RESULTS")
print("=" * 65)

print(f"\nThreshold       : {FINAL_THRESHOLD:.2f}")

print(f"\nAccuracy        : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Macro F1        : {macro_f1:.4f}")
print(f"ROC-AUC         : {roc_auc:.4f}")

print("\nREAL")
print(f"Precision       : {real_precision:.4f}")
print(f"Recall          : {real_recall:.4f}")
print(f"F1              : {real_f1:.4f}")

print("\nFAKE")
print(f"Precision       : {fake_precision:.4f}")
print(f"Recall          : {fake_recall:.4f}")
print(f"F1              : {fake_f1:.4f}")

print("\nCONFUSION MATRIX")
print(f"TN              : {tn}")
print(f"FP              : {fp}")
print(f"FN              : {fn}")
print(f"TP              : {tp}")

print("=" * 65)
```
