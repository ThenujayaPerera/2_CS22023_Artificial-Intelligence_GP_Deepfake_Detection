```python
from tensorflow.keras import layers, models

IMG_SIZE = (128, 128)

base_model_9 = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(128, 128, 3)
)

# Freeze EfficientNet initially
base_model_9.trainable = False

inputs = layers.Input(shape=(128, 128, 3))

# EfficientNetB0
x = base_model_9(inputs, training=False)

# Feature extraction
x = layers.GlobalAveragePooling2D()(x)

# Classification head
x = layers.Dense(256, activation="relu")(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(1, activation="sigmoid")(x)

model_9 = models.Model(inputs, outputs)

model_9.summary()
```

```python
model_9.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

print("Stage 1 model compiled.")
```

```python
# Enable fine-tuning
base_model_9.trainable = True

# Keep early layers frozen
fine_tune_from = 200

for i, layer in enumerate(base_model_9.layers):

    if i < fine_tune_from:
        layer.trainable = False
    else:
        layer.trainable = True

print("Total EfficientNet layers:", len(base_model_9.layers))
print("Fine-tuning starts from layer:", fine_tune_from)

trainable_layers = sum(
    layer.trainable for layer in base_model_9.layers
)

print("Trainable EfficientNet layers:", trainable_layers)
print("Frozen EfficientNet layers:",
      len(base_model_9.layers) - trainable_layers)
```

```python
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score,
    f1_score,
    precision_score,
    recall_score,
    accuracy_score
)

print("Generating test predictions...")

y_true_9 = np.concatenate([
    y.numpy().ravel()
    for _, y in test_ds_9
])

y_prob_9 = model_9.predict(
    test_ds_9,
    verbose=1
).ravel()

print("Number of test labels:", len(y_true_9))
print("Number of predictions:", len(y_prob_9))

print("\nClass mapping:")
print("Fake = 0")
print("Real = 1")
```

```python
threshold_9 = 0.50

y_pred_9 = (
    y_prob_9 >= threshold_9
).astype(int)

cm_9 = confusion_matrix(
    y_true_9,
    y_pred_9
)

print("Confusion Matrix")
print("=" * 50)

print(cm_9)

tn_9, fp_9, fn_9, tp_9 = cm_9.ravel()

print("\nDetailed:")
print("TN:", tn_9)
print("FP:", fp_9)
print("FN:", fn_9)
print("TP:", tp_9)
```

```python
accuracy_9 = accuracy_score(
    y_true_9,
    y_pred_9
)

precision_9 = precision_score(
    y_true_9,
    y_pred_9
)

recall_9 = recall_score(
    y_true_9,
    y_pred_9
)

f1_9 = f1_score(
    y_true_9,
    y_pred_9
)

auc_9 = roc_auc_score(
    y_true_9,
    y_prob_9
)

print("=" * 60)
print("EXPERIMENT 9 TEST METRICS")
print("=" * 60)

print(f"Accuracy  : {accuracy_9:.4f}")
print(f"Precision : {precision_9:.4f}")
print(f"Recall    : {recall_9:.4f}")
print(f"F1 Score  : {f1_9:.4f}")
print(f"ROC-AUC   : {auc_9:.4f}")

print("\nConfusion Matrix")
print("-" * 30)
print(f"TN = {tn_9}")
print(f"FP = {fp_9}")
print(f"FN = {fn_9}")
print(f"TP = {tp_9}")
```

```python
threshold_results_9 = []

for threshold in np.arange(0.20, 0.81, 0.05):

    y_pred = (
        y_prob_9 >= threshold
    ).astype(int)

    cm = confusion_matrix(
        y_true_9,
        y_pred
    )

    tn, fp, fn, tp = cm.ravel()

    acc = accuracy_score(
        y_true_9,
        y_pred
    )

    precision = precision_score(
        y_true_9,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true_9,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true_9,
        y_pred,
        zero_division=0
    )

    threshold_results_9.append([
        threshold,
        acc,
        precision,
        recall,
        f1,
        tn,
        fp,
        fn,
        tp
    ])

threshold_df_9 = pd.DataFrame(
    threshold_results_9,
    columns=[
        "Threshold",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "TN",
        "FP",
        "FN",
        "TP"
    ]
)

display(threshold_df_9)
```
