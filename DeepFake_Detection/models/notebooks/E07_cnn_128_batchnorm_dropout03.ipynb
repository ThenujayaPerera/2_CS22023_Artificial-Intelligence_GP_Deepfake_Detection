```python
import kagglehub

dataset_path = kagglehub.dataset_download(
    "manjilkarki/deepfake-and-real-images"
)

DATASET_DIR = os.path.join(dataset_path, "Dataset")

print("Dataset:")
print(DATASET_DIR)
```

```python
print(os.listdir(DATASET_DIR))
```

```python
IMG_SIZE = (128, 128)
BATCH_SIZE = 32
SEED = 42
```

```python
train_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_DIR, "Train"),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True,
    seed=SEED
)

validation_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_DIR, "Validation"),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

test_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_DIR, "Test"),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)
```

```python
print("Class names:", train_ds_128.class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds_128 = train_ds_128.prefetch(AUTOTUNE)
validation_ds_128 = validation_ds_128.prefetch(AUTOTUNE)
test_ds_128 = test_ds_128.prefetch(AUTOTUNE)

val_ds_128 = validation_ds_128
```

```python
print("TensorFlow version:", tf.__version__)

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print("GPU detected:")
    for gpu in gpus:
        print(gpu)
else:
    print("WARNING: No GPU detected!")
```

```python
tf.keras.utils.set_random_seed(42)
```

```python
from tensorflow.keras import layers, models

model_128_bn_dropout = models.Sequential([

    layers.Input(shape=(128, 128, 3)),

    layers.Rescaling(1./255),

    # Block 1
    layers.Conv2D(
        32,
        (3, 3),
        padding="same"
    ),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(
        64,
        (3, 3),
        padding="same"
    ),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Block 3
    layers.Conv2D(
        128,
        (3, 3),
        padding="same"
    ),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Block 4
    layers.Conv2D(
        256,
        (3, 3),
        padding="same"
    ),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Feature aggregation
    layers.GlobalAveragePooling2D(),

    # Dense classifier
    layers.Dense(256),
    layers.BatchNormalization(),
    layers.Activation("relu"),

    # Experiment 7 change
    layers.Dropout(0.3),

    # Output
    layers.Dense(
        1,
        activation="sigmoid"
    )
])

model_128_bn_dropout.summary()
```

```python
model_128_bn_dropout.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="binary_crossentropy",

    metrics=[
        tf.keras.metrics.BinaryAccuracy(
            name="accuracy"
        ),

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )
    ]
)
```

```python
checkpoint_7 = tf.keras.callbacks.ModelCheckpoint(
    filepath=os.path.join(
        CHECKPOINT_DIR_7,
        "cnn_128_bn_dropout_best.keras"
    ),
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)
```

```python
early_stopping_7 = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)
```

```python
reduce_lr_7 = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.2,
    patience=2,
    min_lr=1e-7,
    verbose=1
)
```

```python
experiment_config_7 = {
    "experiment": "Experiment 7",
    "model": "128x128 CNN + Batch Normalization + Dropout",
    "input_size": "128x128",
    "dropout": 0.3,
    "batch_size": 32,
    "optimizer": "Adam",
    "initial_learning_rate": 0.001,
    "loss": "binary_crossentropy",
    "max_epochs": 20,
    "early_stopping": True,
    "early_stopping_patience": 5,
    "reduce_lr": True,
    "reduce_lr_factor": 0.2,
    "seed": 42,
    "augmentation": "None"
}

config_path = os.path.join(
    EXP7_DIR,
    "experiment_config.json"
)

with open(config_path, "w") as f:
    json.dump(
        experiment_config_7,
        f,
        indent=4
    )

print("Configuration saved:")
print(config_path)
```

```python
history_128_bn_dropout = model_128_bn_dropout.fit(
    train_ds_128,

    validation_data=val_ds_128,

    epochs=20,

    callbacks=[
        checkpoint_7,
        early_stopping_7,
        reduce_lr_7
    ],

    verbose=1
)
```

```python
history_df_7 = pd.DataFrame(
    history_128_bn_dropout.history
)

history_df_7.insert(
    0,
    "epoch",
    range(1, len(history_df_7) + 1)
)

history_csv_path_7 = os.path.join(
    HISTORY_DIR_7,
    "cnn_128_batchnorm_dropout_history.csv"
)

history_df_7.to_csv(
    history_csv_path_7,
    index=False
)

print("History saved:")
print(history_csv_path_7)
```

```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    ConfusionMatrixDisplay
)

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices('GPU'))
```

```python
import kagglehub

dataset_path = kagglehub.dataset_download(
    "manjilkarki/deepfake-and-real-images"
)

DATASET_DIR = os.path.join(dataset_path, "Dataset")

TEST_DIR = os.path.join(DATASET_DIR, "Test")

IMG_SIZE = (128, 128)
BATCH_SIZE = 32
SEED = 42

print("Dataset:", DATASET_DIR)
print("Test:", TEST_DIR)
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

class_names = test_ds.class_names

print("Class names:", class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

test_ds = test_ds.prefetch(AUTOTUNE)
```

```python
EXP7_MODEL_PATH = (
    "/content/drive/MyDrive/DeepFake_Detection/"
    "experiments/cnn_128_batchnorm_dropout/"
    "checkpoints/cnn_128_bn_dropout_best.keras"
)

model_7 = tf.keras.models.load_model(EXP7_MODEL_PATH)

print("Experiment 7 loaded successfully.")
print("Model:", model_7.name)
```

```python
results_7 = model_7.evaluate(
    test_ds,
    verbose=1,
    return_dict=True
)

print("\n========================================")
print("EXPERIMENT 7 â€” TEST RESULTS")
print("========================================")

for key, value in results_7.items():
    print(f"{key:12s}: {value:.4f}")
```

```python
# True labels
y_true_7 = np.concatenate([
    y.numpy().ravel()
    for _, y in test_ds
])

# Predicted probabilities
y_prob_7 = model_7.predict(
    test_ds,
    verbose=1
).ravel()

# Default threshold = 0.5
y_pred_7 = (y_prob_7 >= 0.5).astype(int)

print("Number of test images:", len(y_true_7))
print("Number of predictions:", len(y_pred_7))
```

```python
cm_7 = confusion_matrix(
    y_true_7,
    y_pred_7
)

print("\nExperiment 7 Confusion Matrix:")
print(cm_7)

tn_7, fp_7, fn_7, tp_7 = cm_7.ravel()

print("\n========================================")
print("EXPERIMENT 7 CONFUSION MATRIX VALUES")
print("========================================")

print("TN:", tn_7)
print("FP:", fp_7)
print("FN:", fn_7)
print("TP:", tp_7)
```

```python
accuracy_7 = accuracy_score(
    y_true_7,
    y_pred_7
)

precision_real_7 = precision_score(
    y_true_7,
    y_pred_7,
    pos_label=1
)

recall_real_7 = recall_score(
    y_true_7,
    y_pred_7,
    pos_label=1
)

f1_real_7 = f1_score(
    y_true_7,
    y_pred_7,
    pos_label=1
)

roc_auc_7 = roc_auc_score(
    y_true_7,
    y_prob_7
)

# IMPORTANT:
# Fake = class 0
fake_recall_7 = tn_7 / (tn_7 + fp_7)

fake_precision_7 = tn_7 / (tn_7 + fn_7)

print("\n========================================")
print("EXPERIMENT 7 â€” COMPLETE TEST METRICS")
print("========================================")

print(f"Accuracy              : {accuracy_7:.4f}")
print(f"Precision (Real)      : {precision_real_7:.4f}")
print(f"Recall (Real)         : {recall_real_7:.4f}")
print(f"F1 (Real)             : {f1_real_7:.4f}")
print(f"ROC-AUC               : {roc_auc_7:.4f}")

print("\n--- FAKE CLASS ---")

print(f"Fake Recall           : {fake_recall_7:.4f}")
print(f"Fake Precision        : {fake_precision_7:.4f}")

print("\n--- CONFUSION MATRIX ---")

print(f"TN (Fake â†’ Fake)      : {tn_7}")
print(f"FP (Fake â†’ Real)      : {fp_7}")
print(f"FN (Real â†’ Fake)      : {fn_7}")
print(f"TP (Real â†’ Real)      : {tp_7}")
```
