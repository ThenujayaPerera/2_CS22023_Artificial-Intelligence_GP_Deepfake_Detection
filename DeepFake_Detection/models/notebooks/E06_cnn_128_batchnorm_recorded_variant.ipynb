# E6 - Recorded 128x128 BatchNorm Variant

Recorded E6 metrics are grouped with E5 in the project report. This notebook preserves the BatchNorm implementation and evaluation cells used for that recorded variant.

**Source:** extracted from the original project notebooks. Original cell outputs are preserved where available.


```python
import os

BASE_DRIVE = "/content/drive/MyDrive/DeepFake_Detection"

EXP5_DIR = os.path.join(
    BASE_DRIVE,
    "experiments",
    "cnn_128_batchnorm"
)

CHECKPOINT_DIR_5 = os.path.join(EXP5_DIR, "checkpoints")
MODEL_DIR_5 = os.path.join(EXP5_DIR, "models")
HISTORY_DIR_5 = os.path.join(EXP5_DIR, "history")
GRAPHS_DIR_5 = os.path.join(EXP5_DIR, "graphs")
EVAL_DIR_5 = os.path.join(EXP5_DIR, "evaluation")

for folder in [
    CHECKPOINT_DIR_5,
    MODEL_DIR_5,
    HISTORY_DIR_5,
    GRAPHS_DIR_5,
    EVAL_DIR_5
]:
    os.makedirs(folder, exist_ok=True)

print(EXP5_DIR)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

model_128_bn = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    layers.Rescaling(1./255),

    layers.Conv2D(32, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(256, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.GlobalAveragePooling2D(),

    layers.Dense(256),
    layers.BatchNormalization(),
    layers.Activation("relu"),

    layers.Dropout(0.5),

    layers.Dense(1, activation="sigmoid")
])

model_128_bn.summary()
```

```python
model_128_bn.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc")
    ]
)
```

```python
checkpoint_5 = tf.keras.callbacks.ModelCheckpoint(
    filepath=os.path.join(
        CHECKPOINT_DIR_5,
        "cnn_128_bn_best.keras"
    ),
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)

early_stopping_5 = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

reduce_lr_5 = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.2,
    patience=2,
    min_lr=1e-7,
    verbose=1
)
```

```python
import os
import kagglehub

path = kagglehub.dataset_download(
    "manjilkarki/deepfake-and-real-images"
)

DATASET_PATH = os.path.join(path, "Dataset")

print("Dataset path:")
print(DATASET_PATH)

print("\nFolders:")
print(os.listdir(DATASET_PATH))
```

```python
import tensorflow as tf

IMG_SIZE_128 = (128, 128)
BATCH_SIZE = 32

train_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_PATH, "Train"),
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True,
    seed=42
)

val_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_PATH, "Validation"),
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

test_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_PATH, "Test"),
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("Classes:", train_ds_128.class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds_128 = train_ds_128.prefetch(AUTOTUNE)
val_ds_128 = val_ds_128.prefetch(AUTOTUNE)
test_ds_128 = test_ds_128.prefetch(AUTOTUNE)
```

```python
print("train_ds_128:", train_ds_128)
print("val_ds_128:", val_ds_128)
print("test_ds_128:", test_ds_128)
```

```python
history_128_bn = model_128_bn.fit(
    train_ds_128,
    validation_data=val_ds_128,
    epochs=20,
    callbacks=[
        checkpoint_5,
        early_stopping_5,
        reduce_lr_5
    ]
)
```

```python
import kagglehub

path = kagglehub.dataset_download(
    "manjilkarki/deepfake-and-real-images"
)

DATASET_PATH = os.path.join(path, "Dataset")

print("Dataset path:")
print(DATASET_PATH)
```

```python
IMG_SIZE_128 = (128, 128)
BATCH_SIZE = 32

test_ds_128 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_PATH, "Test"),
    image_size=IMG_SIZE_128,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

test_ds_128 = test_ds_128.prefetch(tf.data.AUTOTUNE)

print("Test dataset ready.")
```

```python
BEST_MODEL_128_BN = os.path.join(
    CHECKPOINT_DIR_5,
    "cnn_128_bn_best.keras"
)

print("Model exists:", os.path.exists(BEST_MODEL_128_BN))
print(BEST_MODEL_128_BN)
```

```python
model_128_bn_best = tf.keras.models.load_model(
    BEST_MODEL_128_BN
)

print("Model loaded successfully.")
model_128_bn_best.summary()
```

```python
test_results_128_bn = model_128_bn_best.evaluate(
    test_ds_128,
    verbose=1
)

print("Test evaluation completed.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Predictions
y_prob_128_bn = model_128_bn_best.predict(
    test_ds_128,
    verbose=1
).ravel()

y_pred_128_bn = (y_prob_128_bn >= 0.5).astype(int)

# True labels
y_true_128_bn = np.concatenate([
    y.numpy() for x, y in test_ds_128
])

# Metrics
accuracy_128_bn = accuracy_score(
    y_true_128_bn,
    y_pred_128_bn
)

precision_128_bn = precision_score(
    y_true_128_bn,
    y_pred_128_bn
)

recall_128_bn = recall_score(
    y_true_128_bn,
    y_pred_128_bn
)

f1_128_bn = f1_score(
    y_true_128_bn,
    y_pred_128_bn
)

auc_128_bn = roc_auc_score(
    y_true_128_bn,
    y_prob_128_bn
)

cm_128_bn = confusion_matrix(
    y_true_128_bn,
    y_pred_128_bn
)

print("==========================================")
print("128Ã—128 BATCH NORMALIZATION")
print("FINAL TEST RESULTS")
print("==========================================")

print(
    f"Accuracy : {accuracy_128_bn:.4f} "
    f"({accuracy_128_bn*100:.2f}%)"
)

print(
    f"Precision: {precision_128_bn:.4f} "
    f"({precision_128_bn*100:.2f}%)"
)

print(
    f"Recall   : {recall_128_bn:.4f} "
    f"({recall_128_bn*100:.2f}%)"
)

print(f"F1 Score : {f1_128_bn:.4f}")

print(f"ROC-AUC  : {auc_128_bn:.4f}")

print("\nClassification Report:")

print(
    classification_report(
        y_true_128_bn,
        y_pred_128_bn,
        target_names=["Fake", "Real"]
    )
)

print("\nConfusion Matrix:")

print(cm_128_bn)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
import pandas as pd
import numpy as np

thresholds = np.arange(0.20, 0.81, 0.05)

threshold_results = []

for threshold in thresholds:

    y_pred = (y_prob_128_bn >= threshold).astype(int)

    acc = accuracy_score(y_true_128_bn, y_pred)
    precision = precision_score(y_true_128_bn, y_pred)
    recall = recall_score(y_true_128_bn, y_pred)
    f1 = f1_score(y_true_128_bn, y_pred)

    cm = confusion_matrix(
        y_true_128_bn,
        y_pred
    )

    tn, fp, fn, tp = cm.ravel()

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Accuracy": acc,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

threshold_df = pd.DataFrame(threshold_results)

print(threshold_df.to_string(index=False))
```
