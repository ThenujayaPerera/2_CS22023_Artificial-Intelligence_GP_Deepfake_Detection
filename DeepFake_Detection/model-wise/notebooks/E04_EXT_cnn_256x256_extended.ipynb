```python
import os

DATASET_PATH = os.path.join(path, "Dataset")

print(os.listdir(DATASET_PATH))
```

```python
import tensorflow as tf

IMG_SIZE_256 = (256, 256)
BATCH_SIZE = 32

test_ds_256 = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATASET_PATH, "Test"),
    image_size=IMG_SIZE_256,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print(test_ds_256.class_names)
```

```python
BEST_MODEL_256_EXT = (
    "/content/drive/MyDrive/DeepFake_Detection/"
    "experiments/cnn_256_extended/"
    "checkpoints/cnn_256_ext_best.keras"
)

print(os.path.exists(BEST_MODEL_256_EXT))
```

```python
model_256_ext_best = tf.keras.models.load_model(
    BEST_MODEL_256_EXT
)

model_256_ext_best.summary()
```

```python
test_results_256_ext = model_256_ext_best.evaluate(
    test_ds_256,
    verbose=1
)

print(test_results_256_ext)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

import numpy as np

# Predictions
y_prob_256_ext = model_256_ext_best.predict(
    test_ds_256,
    verbose=1
).ravel()

# Convert probabilities to class predictions
y_pred_256_ext = (
    y_prob_256_ext >= 0.5
).astype(int)

# True labels
y_true_256_ext = np.concatenate([
    y.numpy()
    for x, y in test_ds_256
])

# Metrics
accuracy_256_ext = accuracy_score(
    y_true_256_ext,
    y_pred_256_ext
)

precision_256_ext = precision_score(
    y_true_256_ext,
    y_pred_256_ext
)

recall_256_ext = recall_score(
    y_true_256_ext,
    y_pred_256_ext
)

f1_256_ext = f1_score(
    y_true_256_ext,
    y_pred_256_ext
)

auc_256_ext = roc_auc_score(
    y_true_256_ext,
    y_prob_256_ext
)

print("==========================================")
print("256Ã—256 EXTENDED TEST RESULTS")
print("==========================================")

print(
    f"Accuracy : {accuracy_256_ext:.4f} "
    f"({accuracy_256_ext*100:.2f}%)"
)

print(
    f"Precision: {precision_256_ext:.4f} "
    f"({precision_256_ext*100:.2f}%)"
)

print(
    f"Recall   : {recall_256_ext:.4f} "
    f"({recall_256_ext*100:.2f}%)"
)

print(
    f"F1 Score : {f1_256_ext:.4f}"
)

print(
    f"ROC-AUC  : {auc_256_ext:.4f}"
)

print("\nClassification Report:")

print(
    classification_report(
        y_true_256_ext,
        y_pred_256_ext,
        target_names=["Fake", "Real"]
    )
)

cm_256_ext = confusion_matrix(
    y_true_256_ext,
    y_pred_256_ext
)

print("\nConfusion Matrix:")
print(cm_256_ext)
```

```python
import json
import os

EVAL_DIR_256_EXT = os.path.join(
    EXT_DIR,
    "evaluation"
)

os.makedirs(EVAL_DIR_256_EXT, exist_ok=True)

metrics_256_ext = {
    "experiment": "cnn_256_extended",
    "input_size": "256x256",
    "best_epoch": 12,
    "accuracy": float(accuracy_256_ext),
    "precision": float(precision_256_ext),
    "recall": float(recall_256_ext),
    "f1_score": float(f1_256_ext),
    "roc_auc": float(auc_256_ext),
    "test_samples": int(len(y_true_256_ext))
}

metrics_path = os.path.join(
    EVAL_DIR_256_EXT,
    "final_metrics.json"
)

with open(metrics_path, "w") as f:
    json.dump(
        metrics_256_ext,
        f,
        indent=4
    )

print("Saved:")
print(metrics_path)
```
